# A fluent fake proof

Let $n$ points be given such that the distance between any two of them is strictly less than $x$.
Show that all the points lie inside a disk of diameter $x$.

(this is false, cf https://en.wikipedia.org/wiki/Jung%27s_theorem)

- 2025: ChatGPT "proved" it, with Thales' theorem.
- 2026: two free chatbots out of three say it is false, give a counterexample and Jung's theorem.
  The third one writes a new proof.

# The new proof (October 2026)

> **2. Show Pairwise Intersection.** [...] every pair of disks $D_i$ and $D_j$ intersects (they overlap).
>
> **3. Apply Helly's Theorem.** A classic theorem in geometry, Helly's Theorem, states that a family of
> convex sets in the plane (in this case, our closed disks) has a non-empty common intersection if every
> pair of sets in the family intersects.
>
> **4. Conclusion.** [...] they all fit inside a disk of diameter $x$. $\blacksquare$

Helly's theorem, in the plane, needs every **three** sets to intersect, not every two.
Counterexample: an equilateral triangle of side $0.99x$ needs a disk of diameter $2 \times 0.99x/\sqrt{3} \approx 1.14x$.

Fluent, confident, makes you want to buy it. But dead wrong at one precise step that you can check: look up the theorem, read its
hypotheses. That is what *Check logic* means in Part IIa.

# Security

# Intermezzo: zoom on chatbots

To become a true "chatbot":
- Step #1, train a "word generator"
  - Enormous amount of data
  - Months of training
- Step #2, *fine-tuning*: learn roles (system, user, assistant)
  - Custom-made quality examples
  - Days of training
- Step #3, during inference, mix the instructions and user input

To become a true "agent":
- Step #4 (agents, Part IIa): give it tools, and run it in a loop

# Security: What could go wrong?

- Data poisoning (LLMs scrape everything, you can use this)

- Prompt engineering (a.k.a. forcing the LLM to do what you want)

- Prompt injection (anything the model reads can become an instruction)

- Agents (the model acts, with *your* permissions)

# Data poisoning



About 250 poisoned documents were enough to plant a backdoor during training, in models of every size
tested (600M to 13B parameters), whatever the amount of clean data: 0.00016% of the training tokens for
the largest ([arXiv:2510.07192](https://arxiv.org/abs/2510.07192), Oct. 2025). The figure shows the same
attack during fine-tuning: a meaningless trigger phrase unlocks harmful answers.

# Poisoning at scale: LLM grooming

  the target is the crawlers that feed the LLMs.
- Ten leading chatbots repeated its false narratives 33% of the time
  ([NewsGuard, March 2025](https://www.newsguardtech.com/special-reports/moscow-based-global-news-network-infected-western-artificial-intelligence-russian-propaganda/)).

# Prompt injection

- The model sees one stream of text: your instructions, and everything it reads (a web page, a README,
  an issue, an e-mail, a PDF). It cannot reliably tell them apart.
- Number one risk of the [OWASP Top 10 for LLM applications](https://genai.owasp.org/llmrisk/llm01-prompt-injection/),
  2025 and 2026 editions.
- With a chatbot, an injection gives a wrong answer. With an agent, it gives an action.

In Part IIa, our small local agent ignored the instruction hidden in a README. Do not generalize: the
attacks you will try in Agent Breaker (TME2, exercise 5) target strong models.

# The lethal trifecta (a.k.a. the death combo)

An agent that has all three:

1. Access to your private data;
2. Exposure to untrusted content;
3. The ability to externally communicate.

> "If your agent combines these three features, an attacker can easily trick it into accessing your
> private data and sending it to that attacker." ([Simon Willison, June 2025](https://simonwillison.net/2025/Jun/16/the-lethal-trifecta/))

Remove one leg.

# Security: what to do

- Run agents in a sandbox (a folder, a container), never with production access.
- Never give one agent the three legs of the trifecta.
- Auto-approve modes ("YOLO", "Turbo"...) are for disposable environments.
- Before installing a package an AI suggested, check that it exists, and that it is the one you think.
- Everything your agent reads (web pages, READMEs, issues, e-mails) is untrusted input.

# Ecology: the unit is now the task

- A long question with a reasoning model: around 10 to 35 Wh (estimates, [Jegham et al.](https://arxiv.org/abs/2505.09598)).

  hundreds of times a chat message.
- Why: at each step, the agent rereads its whole context.
- The frontier agent of Part IIa worked 32 minutes and made 147 tool calls for one question: probably
  thousands of chat messages, around one kilowatt-hour.

The last line is our own order of magnitude (0.25 to 2 kWh), from the per-call figures of the two
sources above. Nobody publishes the energy of one agent call. A dishwasher cycle is about 1 kWh.

# Conclusion: what the human adds

# What you add

What the tool cannot sign for you:

- **Choose**: what to delegate, and what to learn first.
- **Specify**: say what counts as a solution.
- **Check**: run the code, read it, debug it, estimate an order of magnitude, open the source.
- **Own**: you sign it. The reviewer is not your verifier.

A team does not hire someone to forward prompts. It hires someone it can trust to check.
